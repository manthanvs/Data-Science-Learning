## Objective

To analyze student grades using Streamlit by selecting different subjects and visualizing their performance using statistical measures and charts.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("grades.csv")
df.head()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print(df["Subject"].unique())
```

```python
selected_subject = "Math"

subject_data = df[df["Subject"] == selected_subject]

subject_data.head()
```

```python
mean_marks = subject_data["Final"].mean()
median_marks = subject_data["Final"].median()
std_marks = subject_data["Final"].std()

print("Mean :", round(mean_marks,2))
print("Median :", round(median_marks,2))
print("Standard Deviation :", round(std_marks,2))
```

```python
plt.figure(figsize=(6,5))

plt.boxplot(subject_data["Final"])

plt.title("Final Marks Distribution")

plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(
    subject_data["Test1"],
    subject_data["Final"]
)

plt.title("Test1 vs Final Marks")

plt.xlabel("Test1")

plt.ylabel("Final")

plt.show()
```
