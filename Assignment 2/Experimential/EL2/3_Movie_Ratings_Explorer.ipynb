```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("movies.csv")
```

```python
df.head()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
df.info()
```

```python
df.isnull().sum()
```

```python
genres = sorted(df["Genre"].unique())
print("Available Genres:")
print(genres)
```

```python
selected_genre = "Action"
genre_df = df[df["Genre"] == selected_genre]
genre_df.head()
```

```python
average_rating = genre_df["Rating"].mean()
average_votes = genre_df["Votes"].mean()
median_year = genre_df["Year"].median()
print("Average Rating :", round(average_rating, 2))
print("Average Votes :", round(average_votes, 2))
print("Median Release Year :", median_year)
```

```python
plt.figure(figsize=(7,5))
plt.boxplot(genre_df["Rating"])
plt.title(f"Ratings Distribution ({selected_genre})")
plt.ylabel("Rating")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    genre_df["Votes"],
    genre_df["Rating"],
    s=genre_df["Votes"] / 50,
    c=genre_df["Rating"],
)
plt.title(f"{selected_genre} Movies")
plt.xlabel("Votes")
plt.ylabel("Rating")
plt.colorbar(label="Rating")
plt.show()
```
