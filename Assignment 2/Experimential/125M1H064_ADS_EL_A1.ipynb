```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Question 1: Analyze Customer Purchase Patterns

Analyze Customer Purchase Patterns: Given a CSV file containing customer transactions (customer_id, purchase_amount, date), use Pandas to calculate the average monthly spending per customer.


```python
import pandas as pd

# Read CSV file
df = pd.read_csv("ExpLearningCSVfiles/customer.csv")

# Convert date column into date format
df["date"] = pd.to_datetime(df["date"])

# Create month column
df["month"] = df["date"].dt.month

# Find average monthly spending of each customer
result = df.groupby("customer_id")["purchase_amount"].mean()

print(result)
```

Question 2: Identify Missing Values and Impute

Using a dataset with missing values (e.g., age, salary), use Pandas and NumPy to identify missing values and fill them with the column mean.

```python

# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/employee.csv")

# Check missing values
print(df.isnull().sum())

# Fill missing values with column mean
df["age"] = df["age"].fillna(df["age"].mean())
df["salary"] = df["salary"].fillna(df["salary"].mean())

print(df)
```

Question 3: Filter and Sort Employee Records

Given employee data (name, department, salary), use Pandas to filter employees in the IT department earning above 60K and sort them by salary.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/employees.csv")

# Filter employees
data = df[(df["department"] == "IT") & (df["salary"] > 60000)]

# Sort by salary
data = data.sort_values(by="salary")

print(data)
```

Question 4: Calculate Statistical Metrics for Stock Data

Using NumPy and Pandas, calculate the mean, median, and standard deviation of daily stock prices from a dataset.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/stocks.csv")

# Calculate statistics
mean = np.mean(df["price"])
median = np.median(df["price"])
std = np.std(df["price"])

print("Mean =", mean)
print("Median =", median)
print("Standard Deviation =", std)
```

Question 5: Create a Pivot Table for Sales Analysis

From a sales dataset (region, product, sales), use Pandas to create a pivot table showing total sales per region for each product.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/sales.csv")

# Create pivot table
table = pd.pivot_table(df,
                       values="sales",
                       index="region",
                       columns="product",
                       aggfunc="sum")

print(table)
```

Question 6: Plot Monthly Sales Using Matplotlib

From a dataset with month and sales, use Matplotlib to create a line chart showing monthly sales trends.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/monthlySales.csv")

# Line chart
plt.plot(df["month"], df["sales"], marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Question 7: Compare Category Sales with Bar Chart

Create a bar chart using Matplotlib to compare total sales for different product categories.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/catagory_sales.csv")

# Bar chart
plt.bar(df["category"], df["sales"])

plt.title("Category Sales")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

Question 8: Visualize Correlation Using Heatmap (Seaborn)

Given a dataset with multiple numeric columns, use Seaborn to draw a heatmap of the correlation matrix.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/data.csv")

# Correlation matrix
corr = df.corr(numeric_only=True)

# Heatmap
sns.heatmap(corr, annot=True)

plt.show()
```

Question 9: Visualize Age Distribution Using Histogram

Using a dataset with an age column, plot a histogram with Seaborn to show the age distribution.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/people.csv")

# Display first 5 rows
print(df.head())

# Plot histogram of age
sns.histplot(df["age"], bins=8)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of People")

plt.show()
```

Question 10: Scatter Plot to Show Relationship Between Two Variables

Use Seaborn to create a scatter plot showing the relationship between advertising budget and sales.

```python
# Read dataset
df = pd.read_csv("ExpLearningCSVfiles/advertising.csv")

# Scatter plot
sns.scatterplot(x="advertising_budget", y="sales", data=df)

plt.title("Advertising Budget vs Sales")
plt.xlabel("Advertising Budget")
plt.ylabel("sales")
plt.show()
```
